### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Ingreso de la tienda



```python
# Calcular ingresos totales por tienda

ingreso_tienda1 = tienda["Precio"].sum()
ingreso_tienda2 = tienda2["Precio"].sum()
ingreso_tienda3 = tienda3["Precio"].sum()
ingreso_tienda4 = tienda4["Precio"].sum()

print("Ingresos Tienda 1:", ingreso_tienda1)
print("Ingresos Tienda 2:", ingreso_tienda2)
print("Ingresos Tienda 3:", ingreso_tienda3)
print("Ingresos Tienda 4:", ingreso_tienda4)
```

```python
import matplotlib.pyplot as plt

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
ingresos = [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]

plt.bar(tiendas, ingresos)
plt.title("Ingresos por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")
plt.show()
```

# 2. Ventas por categoría

```python
# Ventas por categoría en cada tienda

categorias_tienda1 = tienda["Categoría del Producto"].value_counts()
categorias_tienda2 = tienda2["Categoría del Producto"].value_counts()
categorias_tienda3 = tienda3["Categoría del Producto"].value_counts()
categorias_tienda4 = tienda4["Categoría del Producto"].value_counts()

print("Ventas por categoría - Tienda 1")
print(categorias_tienda1)

print("\nVentas por categoría - Tienda 2")
print(categorias_tienda2)

print("\nVentas por categoría - Tienda 3")
print(categorias_tienda3)

print("\nVentas por categoría - Tienda 4")
print(categorias_tienda4)
```

```python
import matplotlib.pyplot as plt

# Contar ventas por categoría en cada tienda
categorias_tienda1 = tienda["Categoría del Producto"].value_counts()
categorias_tienda2 = tienda2["Categoría del Producto"].value_counts()
categorias_tienda3 = tienda3["Categoría del Producto"].value_counts()
categorias_tienda4 = tienda4["Categoría del Producto"].value_counts()

# Crear una figura con 4 gráficos
plt.figure(figsize=(12,10))

# Tienda 1
plt.subplot(2,2,1)
plt.pie(categorias_tienda1, labels=categorias_tienda1.index, autopct="%1.1f%%")
plt.title("Tienda 1")

# Tienda 2
plt.subplot(2,2,2)
plt.pie(categorias_tienda2, labels=categorias_tienda2.index, autopct="%1.1f%%")
plt.title("Tienda 2")

# Tienda 3
plt.subplot(2,2,3)
plt.pie(categorias_tienda3, labels=categorias_tienda3.index, autopct="%1.1f%%")
plt.title("Tienda 3")

# Tienda 4
plt.subplot(2,2,4)
plt.pie(categorias_tienda4, labels=categorias_tienda4.index, autopct="%1.1f%%")
plt.title("Tienda 4")

plt.tight_layout()
plt.show()


```

# 3. Calificación promedio de la tienda


```python
# Calificación promedio de clientes por tienda

promedio_tienda1 = tienda["Calificación"].mean()
promedio_tienda2 = tienda2["Calificación"].mean()
promedio_tienda3 = tienda3["Calificación"].mean()
promedio_tienda4 = tienda4["Calificación"].mean()

print("Calificación promedio Tienda 1:", promedio_tienda1)
print("Calificación promedio Tienda 2:", promedio_tienda2)
print("Calificación promedio Tienda 3:", promedio_tienda3)
print("Calificación promedio Tienda 4:", promedio_tienda4)
```

```python
import matplotlib.pyplot as plt

# Calificación promedio por tienda
promedio_tienda1 = tienda["Calificación"].mean()
promedio_tienda2 = tienda2["Calificación"].mean()
promedio_tienda3 = tienda3["Calificación"].mean()
promedio_tienda4 = tienda4["Calificación"].mean()

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
calificaciones = [promedio_tienda1, promedio_tienda2, promedio_tienda3, promedio_tienda4]

plt.figure()

plt.plot(tiendas, calificaciones, marker="o")

plt.title("Calificación promedio de clientes por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Calificación promedio")

plt.show()
```

# 4. Productos más y menos vendidos

```python
# Productos más y menos vendidos por tienda

ventas_tienda1 = tienda["Producto"].value_counts()
ventas_tienda2 = tienda2["Producto"].value_counts()
ventas_tienda3 = tienda3["Producto"].value_counts()
ventas_tienda4 = tienda4["Producto"].value_counts()

print("Tienda 1")
print("Más vendido:", ventas_tienda1.idxmax())
print("Menos vendido:", ventas_tienda1.idxmin())

print("\nTienda 2")
print("Más vendido:", ventas_tienda2.idxmax())
print("Menos vendido:", ventas_tienda2.idxmin())

print("\nTienda 3")
print("Más vendido:", ventas_tienda3.idxmax())
print("Menos vendido:", ventas_tienda3.idxmin())

print("\nTienda 4")
print("Más vendido:", ventas_tienda4.idxmax())
print("Menos vendido:", ventas_tienda4.idxmin())
```

```python
import folium
from folium.plugins import HeatMap
import pandas as pd

# Unir las coordenadas de todas las tiendas
coordenadas = pd.concat([
    tienda[["lat","lon"]],
    tienda2[["lat","lon"]],
    tienda3[["lat","lon"]],
    tienda4[["lat","lon"]]
])

# Crear mapa centrado en el promedio de coordenadas
mapa = folium.Map(
    location=[coordenadas["lat"].mean(), coordenadas["lon"].mean()],
    zoom_start=4
)

# Crear mapa de calor
HeatMap(coordenadas.values).add_to(mapa)

mapa
```

Con estos datos y los gráficos realizados, se buscó comparar el desempeño de las cuatro tiendas para identificar cuál tiene mejores resultados.

También se analizaron las calificaciones promedio de los clientes. La tienda con mejor calificación es la Tienda 3, con un promedio de 4.04, seguida de la Tienda 2, la Tienda 4 y finalmente la Tienda 1.

Otro punto que se revisó fue los productos más y menos vendidos en cada tienda. Por ejemplo, en la Tienda 1 el producto más vendido es el microondas y el menos vendido son los auriculares con micrófono. Después de analizar todos los datos, se puede concluir que la Tienda 1 es una buena opción para vender, ya que tiene los ingresos más altos entre todas las tiendas.
